```python
import pandas as pd

# Load the dataset
df = pd.read_excel('Online retail.xlsx', header=None)

# Rename column for clarity
df.columns = ['Items']

# Preview the data
df.head()
df.info()
```

```python
# Split items into lists
df['Items'] = df['Items'].apply(lambda x: x.split(','))

# Remove duplicates (if needed)
df.drop_duplicates(inplace=True)

# One-hot encoding using TransactionEncoder
from mlxtend.preprocessing import TransactionEncoder

te = TransactionEncoder()
te_array = te.fit(df['Items']).transform(df['Items'])
basket_df = pd.DataFrame(te_array, columns=te.columns_)
```

```python
# Step 1: Install mlxtend (only once per runtime)
!pip install mlxtend

# Step 2: Import the functions
from mlxtend.frequent_patterns import apriori, association_rules

# Step 3: Now use apriori
frequent_itemsets = apriori(basket_df, min_support=0.05, use_colnames=True)

rules = association_rules(frequent_itemsets, metric="lift", min_threshold=1)
filtered_rules = rules[(rules['confidence'] >= 0.4) & (rules['lift'] >= 1.1)]

top_rules = filtered_rules.sort_values(by='lift', ascending=False).head(10)
print(top_rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']])
```

## Interpretation
- High lift values (>1.2) indicate strong associations between products.
- These rules suggest which items are frequently bought together.
- Businesses can use these insights for product bundling, cross-selling, and inventory planning.

## Interview Questions

**1. What is Lift and Why is it Important?**  
Lift measures how much more likely items are purchased together than expected if independent. It helps identify strong, non-random associations.

**2. What is Support and Confidence? How Are They Calculated?**  
- Support = (Transactions containing itemset) / (Total transactions)  
- Confidence = (Transactions containing both antecedent and consequent) / (Transactions containing antecedent)

**3. Limitations of Association Rule Mining**  
- Scalability: Can be slow on large datasets  
- Interpretability: Too many rules can overwhelm users  
- Relevance: Not all rules are actionable or useful
