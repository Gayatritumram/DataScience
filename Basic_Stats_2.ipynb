```python
import numpy as np
data = [1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29]
sample_mean = np.mean(data)
```

```python
sample_std = np.std(data, ddof=1)
```

```python
n = len(data)
```

```python
from scipy.stats import t
t_value = t.ppf(0.995, df=n-1)  # 99% CI -> 0.995 for two-tailed
```

```python
MoE = t_value * (sample_std / np.sqrt(n))
```

```python
CI_lower = sample_mean - MoE
CI_upper = sample_mean + MoE
(CI_lower, CI_upper)
```

```python
from scipy.stats import norm
z_value = norm.ppf(0.995)  # 99% CI -> 0.995 for two-tailed
```

```python
MoE_known_sigma = z_value * (0.2 / np.sqrt(n))
```

```python
CI_lower_known_sigma = sample_mean - MoE_known_sigma
CI_upper_known_sigma = sample_mean + MoE_known_sigma
(CI_lower_known_sigma, CI_upper_known_sigma)
```
