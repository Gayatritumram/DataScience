1. State the Hypotheses:
Null Hypothesis (H₀): There is no significant association between the type of smart home device purchased and customer satisfaction levels.

Alternative Hypothesis (H₁): There is a significant association between the type of smart home device purchased and customer satisfaction levels.

2. Compute the Chi-Square Statistic:
Here's a Python code to perform the Chi-Square test:

# CHI-SQUARE TEST

```python
import numpy as np
import pandas as pd
from scipy.stats import chi2_contingency

# Data
data = {
    "Satisfaction": ["Very Satisfied", "Satisfied", "Neutral", "Unsatisfied", "Very Unsatisfied"],
    "Smart Thermostat": [50, 80, 60, 30, 20],
    "Smart Light": [70, 100, 90, 50, 50]
}

# Create DataFrame
df = pd.DataFrame(data)

# Compute the Chi-Square statistic
observed = df[["Smart Thermostat", "Smart Light"]].values
chi2, p, dof, expected = chi2_contingency(observed)

print(f"Chi-Square Statistic: {chi2}")
print(f"P-Value: {p}")
print(f"Degrees of Freedom: {dof}")
print("Expected Frequencies:")
print(expected)
```

3. Determine the Critical Value:
Using a significance level (alpha) of 0.05 and degrees of freedom (df):

Degrees of Freedom (df):
(rows-1)x(columns-1)=(5-1)x(2-1)=4
Critical Value: You can use a Chi-Square distribution table or a Python function to determine this.

```python
from scipy.stats import chi2

# Significance level
alpha = 0.05

# Critical value for 4 degrees of freedom
critical_value = chi2.ppf(1 - alpha, dof)
print(f"Critical Value: {critical_value}")
```

4. Make a Decision:
Compare the Chi-Square statistic with the critical value to decide whether to reject the null hypothesis

```python
if (chi2>critical_value):
    print("Reject the null hypothesis. There is a significant association between the type of device purchased and customer satisfaction.")
else:
    print("Fail to reject the null hypothesis. There is no significant association between the type of device purchased and customer satisfaction.")
```

```python
import numpy as np
import pandas as pd
from scipy.stats import chi2_contingency, chi2

# Data
data = {
    "Satisfaction": ["Very Satisfied", "Satisfied", "Neutral", "Unsatisfied", "Very Unsatisfied"],
    "Smart Thermostat": [50, 80, 60, 30, 20],
    "Smart Light": [70, 100, 90, 50, 50]
}

# Create DataFrame
df = pd.DataFrame(data)

# Compute the Chi-Square statistic
observed = df[["Smart Thermostat", "Smart Light"]].values
chi2_stat, p, dof, expected = chi2_contingency(observed)

# Determine the Critical Value
alpha = 0.05
critical_value = chi2.ppf(1 - alpha, df=dof)

# Results
print(f"Chi-Square Statistic: {chi2_stat}")
print(f"P-Value: {p}")
print(f"Degrees of Freedom: {dof}")
print(f"Critical Value: {critical_value}")
print("Expected Frequencies:")
print(expected)

# Make a Decision
if chi2_stat > critical_value:
    print("Reject the null hypothesis. There is a significant association between the type of device purchased and customer satisfaction.")
else:
    print("Fail to reject the null hypothesis. There is no significant association between the type of device purchased and customer satisfaction.")
```

Conclusion:
Chi-Square Statistic: The calculated value.

P-Value: The probability of observing the data given that the null hypothesis is true.

Critical Value: The threshold value at which you decide whether to reject the null hypothesis.

Decision: Based on comparing the Chi-Square statistic to the critical value.



# **HYPOTHESIS** **TESTING**

1. State the Hypotheses:
Null Hypothesis (H₀): The weekly operating cost follows the theoretical model
W=1000+5X.

Alternative Hypothesis (H₁): The weekly operating cost is higher than what the theoretical model suggests.

2. Calculate the Test Statistic:
Given:

Sample mean weekly cost (𝑥) = Rs. 3,050

Theoretical mean weekly cost
(𝜇) =100+5×600 = Rs. 4,000

Standard deviation of the sample
(𝜎) = 5×25 = Rs. 125

Sample size
(𝑛) = 25

The test statistic
(𝑡) is calculated using the formula:

𝑡=𝑥−𝜇/𝜎/𝑛


```python
import math

# Given values
sample_mean = 3050
theoretical_mean = 1000 + 5 * 600
sample_std_dev = 5 * 25
sample_size = 25

# Calculate the test statistic
t_statistic = (sample_mean - theoretical_mean) / (sample_std_dev / math.sqrt(sample_size))
t_statistic
```

3. Determine the Critical Value:
Using the alpha level of 5% (𝛼=0.05), we need to determine the critical value from the standard normal (Z) distribution table.

```python
from scipy.stats import norm

# Significance level
alpha = 0.05

# Critical value for one-tailed test at alpha = 0.05
critical_value = norm.ppf(1 - alpha)
critical_value
```

4. Make a Decision:
Compare the test statistic with the critical value:

```python
if t_statistic > critical_value:
    decision = "Reject the null hypothesis. There is a significant increase in the weekly operating costs."
else:
    decision = "Fail to reject the null hypothesis. The weekly operating costs are not significantly higher."
decision
```

5. Conclusion:
Based on the comparison between the test statistic and the critical value, we can conclude whether there is strong evidence to support the restaurant owners' claim that the weekly operating costs are higher than the model suggests.

Full Python Code:
Here's the full Python code detailing each step:

```python
import math
from scipy.stats import norm

# Given values
sample_mean = 3050
theoretical_mean = 1000 + 5 * 600
sample_std_dev = 5 * 25
sample_size = 25

# Calculate the test statistic
t_statistic = (sample_mean - theoretical_mean) / (sample_std_dev / math.sqrt(sample_size))
print(f"Test Statistic: {t_statistic}")

# Significance level
alpha = 0.05

# Critical value for one-tailed test at alpha = 0.05
critical_value = norm.ppf(1 - alpha)
print(f"Critical Value: {critical_value}")

# Make a Decision
if t_statistic > critical_value:
    decision = "Reject the null hypothesis. There is a significant increase in the weekly operating costs."
else:
    decision = "Fail to reject the null hypothesis. The weekly operating costs are not significantly higher."
print(decision)
```

Conclusion:
Test Statistic: The calculated value from the sample data.

Critical Value: The threshold value from the Z distribution table.

Decision: Based on the comparison of the test statistic and the critical value.
